```python
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

import pickle
```

```python
df = pd.read_csv("../../data/pre_data.csv")
```

```python
exog_cols = ["Thị_trường", "Loại_giá", "Nguồn"]
```

```python
df.head()
```

```python
for item in df["Tên_mặt_hàng"].unique():
    item_df = df[df["Tên_mặt_hàng"] == item].sort_values("Ngày")

    y = item_df["Giá"].values
    X = item_df[exog_cols].values

    model = SARIMAX(
        y,
        exog=X,
        order=(3, 2, 2),             
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    results = model.fit(disp=False)

    with open(f"../../models/sarimax/{item}.pkl", "wb") as file:
        pickle.dump(results, file)
```

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
from statsmodels.tsa.statespace.sarimax import SARIMAX

import pickle
```

```python
df = pd.read_csv("../../data/pre_data.csv")
```

```python
exog_cols = ["Thị_trường", "Loại_giá", "Nguồn"]
```

```python
df.head()
```

```python
for item in df["Tên_mặt_hàng"].unique():
    item_df = df[df["Tên_mặt_hàng"] == item]

    y = item_df["Giá"].values
    X = item_df[exog_cols].values

    model = SARIMAX(
        y,
        exog=X,
        order=(1, 1, 1),             
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    with open(f"../../results/sarimax/{item}.pkl", "wb") as file:
        pickle.dump(model, file)
```

```python
import pandas as pd
import pickle
from statsmodels.tsa.statespace.sarimax import SARIMAX

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../../data/pre_data.csv", parse_dates=True)
```

```python
for item in df["Tên_mặt_hàng"].unique():
    item_df = df[df["Tên_mặt_hàng"]==item]

    y = item_df["Giá"]
    X = item_df[["Thị_trường", "Loại_giá", "Nguồn"]]

    model = SARIMAX(
        y,
        exog=X,         
        order=(1,1,1),
        seasonal_order=(0,0,0,0),
    )

    results = model.fit()

    with open(f"../../models/sarimax/{item}.pkl", "wb") as file:
        pickle.dump(results, file)
```

```python
forecast = results.get_forecast(steps=1, exog=[[0, 0, 0]])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("forecast_from_training.csv")
```

```python
df.head()
```

```python
df = df.sort_values("Ngày")
```

```python
date = df["Ngày"]
actual = df["Actual_Giá"]
forecast = df["Forecasted_Giá"]
```

```python
plt.plot(date, actual, label="y_true")
plt.plot(date, forecast, label="y_pred")
plt.legend()
```

```python
mse = mean_squared_error(actual, forecast)
print(mse)
```
